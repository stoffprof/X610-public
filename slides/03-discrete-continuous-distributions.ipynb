```python
p = 0.3; n = 100_000
waits = np.argmax(rng.random((n, 16)) < p, axis=1) + 1
x = np.arange(1, 15)
fig, ax = plt.subplots(figsize=(7, 3.2))
ax.hist(waits, bins=x - 0.5, density=True, alpha=0.4, label='simulated')
ax.stem(x, stats.geom(p).pmf(x), label=f'Geometric({p}) PMF')
ax.set(xlabel='trials until first success', ylabel='probability')
ax.legend()
plt.show()
```

```python
n, p = 10, 0.3
draws = (rng.random((100_000, n)) < p).sum(axis=1)

x = np.arange(n + 1)
fig, ax = plt.subplots(figsize=(7, 3.2))
ax.hist(draws, bins=np.arange(n + 2) - 0.5, density=True, alpha=0.4,
        label='simulated')
ax.stem(x, stats.binom(n, p).pmf(x), label='Bin(10, 0.3) PMF')
ax.set(xlabel='successes', ylabel='probability')
ax.legend()
plt.show()
```

```python
λ, x = 5, np.arange(16)

fig, ax = plt.subplots(figsize=(7, 3.2))
for n in (10, 30, 1000):
    ax.plot(x, stats.binom(n, λ / n).pmf(x), 'o--', alpha=0.6,
            label=f'Bin({n}, {λ}/{n})')
ax.plot(x, stats.poisson(λ).pmf(x), 'k.-', label='Poisson(5)')
ax.set(xlabel='x', ylabel='f(x)')
ax.legend()
plt.show()
```

```python
waits = rng.exponential(scale=1/0.5, size=100_000)   # lambda = 0.5

x = np.linspace(0, 12, 200)
fig, ax = plt.subplots(figsize=(7, 3.2))
ax.hist(waits, bins=60, density=True, alpha=0.4, label='simulated')
ax.plot(x, stats.expon(scale=2).pdf(x), 'k', label='Exponential(0.5) PDF')
ax.set(xlabel='waiting time', ylabel='density')
ax.set(xlim=(0, 12))
ax.legend()
plt.show()
```

```python
mu, sigma = 0.01, 0.05

# ppf ('percent point function') is SciPy's name for the quantile function
ppf = stats.norm.ppf(0.05, loc=mu, scale=sigma)

# The sample quantile of simulated returns
returns = rng.normal(mu, sigma, size=100_000)
quantile_5 = np.quantile(returns, 0.05)

print(f'Analytic 5th percentile:  {ppf:.4f}')
print(f'Simulated 5th percentile: {quantile_5:.4f}')
```

```python
p = 0.3
z = np.linspace(-4, 4, 400)

fig, ax = plt.subplots(figsize=(7, 3.2))
for n in (5, 25, 50):
    k = np.arange(n + 1)
    zk = (k - n * p) / np.sqrt(n * p * (1 - p))
    ax.plot(zk, stats.binom(n, p).pmf(k) * np.sqrt(n * p * (1 - p)),
            'o', ms=3, alpha=0.6, label=f'n = {n}')
ax.plot(z, stats.norm().pdf(z), 'k:', label=r'$\phi(z; 0, 1)$')
ax.set(xlim=(-4, 4), xlabel='z', ylabel='scaled PMF')
ax.legend()
plt.show()
```

```python
from scipy.special import gamma, factorial

plt.figure(figsize=(7,4))
x = np.linspace(0.05,5.1,200)
plt.plot(x, gamma(x), 'g--')
n = np.arange(6)
plt.plot(n, gamma(n), 'ko')
plt.xlabel('$x$', fontsize=18)
plt.ylabel(r'$\Gamma(x)$', fontsize=18)
plt.grid(ls=':')
plt.show()
```

```python
Z = rng.standard_normal((100_000, 5))
chi2_sim = (Z ** 2).sum(axis=1)                  # chi-square, 5 df

x = np.linspace(0, 20, 300)
fig, ax = plt.subplots(figsize=(7, 3.2))
ax.hist(chi2_sim, bins=80, density=True, alpha=0.4, label='sum of 5 $Z_i^2$')
ax.plot(x, stats.chi2(5).pdf(x), 'k', label=r'$\chi^2_5$ PDF')
ax.set(xlabel='x', ylabel='density')
ax.set_xlim(0, 20)
ax.legend()
plt.show()
```
