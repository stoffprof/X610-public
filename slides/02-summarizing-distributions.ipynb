```python
n = 1_000_000
X = rng.integers(1, 7, size=n)          # die roll
Y = X + rng.exponential(size=n)         # strongly dependent on X

print('E[2X + 3Y + 1] ≈', (2 * X + 3 * Y + 1).mean())
print('2E[X]+3E[Y]+1  ≈', 2 * X.mean() + 3 * Y.mean() + 1)
```

```python
X = stats.randint(1, 7)  # fair die: values 1 through 6
mean = X.moment(1)
second_moment = X.moment(2)

print('E[X] =', mean)
print('E[X²] =', second_moment)
print('V[X] from deviations =', X.expect(lambda x: (x - mean) ** 2))
print('V[X] from moments =', second_moment - mean ** 2)
```

```python
tight = stats.rv_discrete(values=([-2, 0, 2], [1/8, 3/4, 1/8]))
dists = {'normal': stats.norm(), 'uniform': stats.uniform(),
         'exponential': stats.expon(), 'three-point': tight}

print('eps=2: bound = 0.25')
for name, d in dists.items():
    m, s = d.mean(), d.std()
    actual = d.expect(lambda x: np.abs(x - m) >= 2 * s)
    print(f'  {name:12s} actual = {actual:.4f}')
```

```python
X = rng.exponential(size=200_000)       # skewed: E[X]=1, median≈0.69

cs = np.linspace(0.2, 2.0, 100)
mse = [((X - c) ** 2).mean() for c in cs]
print('argmin MSE ≈', cs[np.argmin(mse)], '   E[X] ≈', X.mean())
```

```python
X = rng.integers(-1, 2, size=1_000_000)
Y = X ** 2

print('corr(X, Y)     ≈', np.corrcoef(X, Y)[0, 1].round(4))
print('E[Y | X = 0]   =', Y[X == 0].mean(), '  E[Y | X = ±1] =', Y[X != 0].mean())
```

```python
n = 1_000_000
X = (rng.random(n) < 0.5).astype(int)                    # coin
Y = np.where(X == 0, rng.integers(1, 5, n), rng.integers(1, 7, n))

print('E[Y]              ≈', Y.mean())
print('E[E[Y|X]] = ½·5/2 + ½·7/2 =', 0.5 * 2.5 + 0.5 * 3.5)
```

```python
n = 2_000
X = rng.random(n)                       # U(0,1)
Y = 10 * X ** 2 + rng.standard_normal(n)

b = np.cov(X, Y)[0, 1] / X.var()        # sample analogs of beta, alpha
a = Y.mean() - b * X.mean()

xs = np.linspace(0, 1, 100)
fig, ax = plt.subplots(figsize=(7, 3.4))
ax.plot(X, Y, '.', ms=2, alpha=0.3)
ax.plot(xs, 10 * xs ** 2, 'k', lw=2, label=r'CEF: $10x^2$')
ax.plot(xs, a + b * xs, 'r', lw=2, label=f'BLP: {a:.2f} + {b:.2f}x')
ax.set(xlabel='x', ylabel='y')
ax.legend()
plt.show()
```
