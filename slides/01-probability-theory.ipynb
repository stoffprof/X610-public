```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

rng = np.random.default_rng(610)
```

```python
flips = rng.integers(0, 2, size=100_000)   # 1 = heads, 0 = tails
running_freq = np.cumsum(flips) / np.arange(1, flips.size + 1)

print(running_freq[[9, 99, 999, 9_999, 99_999]])
```

```python
rolls = rng.integers(1, 7, size=100_000)

A = rolls >= 4          # Boolean value that is True if roll>=4
B = rolls % 2 == 0      # Boolean value that is True if roll is even

print('P(A∩B) ≈', (A & B).mean())            # 1/3
print('P(A∪B) ≈', (A | B).mean())            # 1/2 + 1/2 - 1/3 = 2/3
print('check  :', A.mean() + B.mean() - (A & B).mean())
```

```python
print('P(A|B) ≈', A[B].mean())      # restrict to draws where B occurred
print('exact  :', (2/6) / (3/6))    # P(A∩B)/P(B) = 2/3
```

```python
n = 100_000
heads = rng.random(n) < 0.5
die = np.where(heads,
               rng.integers(1, 5, size=n),    # 4-sided if heads
               rng.integers(1, 7, size=n))    # 6-sided if tails

three = die == 3
print('P(heads | rolled 3) ≈', heads[three].mean())
```

```python
n = 1_000_000
sick = rng.random(n) < 0.01
positive = np.where(sick, rng.random(n) < 0.95, rng.random(n) < 0.05)

print('P(sick | +) ≈', sick[positive].mean())
```

```python
#| code-line-numbers: "1|2|5|1-7"
X = stats.binom(n=10, p=0.3)    # the r.v.
x = np.arange(11)               # x = [0, 1, 2, ..., 10]

fig, ax = plt.subplots(figsize=(7, 3))
ax.stem(x, X.pmf(x))            # plot (x,y)
ax.set(xlabel='x', ylabel='f(x)', title='Binomial(10, 0.3) PMF')
plt.show()
```

```python
#| code-line-numbers: "1|2|5|1-7"
#| output-location: fragment
die = stats.randint(1, 7)         # create the r.v.
x = np.linspace(-0.5, 7.5, 500)   # 500 points between -0.5 and 7.5

fig, ax = plt.subplots(figsize=(7, 3))
ax.step(x, die.cdf(x), where='post')    # create the step function
ax.set(xlabel='x', ylabel='F(x)', title='CDF of a fair die roll')
plt.show()
```

```python
n = 100_000
X = (rng.random(n) < 0.5).astype(int)             # 1 = tails
Y = np.where(X == 0, rng.integers(1, 5, size=n),
                     rng.integers(1, 7, size=n))

print('joint    f(0,3)      ≈', ((X == 0) & (Y == 3)).mean())   # 1/8
print('marginal f_Y(3)      ≈', (Y == 3).mean())                # 5/24
print('cond.    f_X|Y(0|3)  ≈', (X[Y == 3] == 0).mean())        # 3/5
```
